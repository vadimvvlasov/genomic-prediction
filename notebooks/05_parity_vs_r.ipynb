## Структура сравнения

Когда R будет доступен, нужно:

1. **Запустить R-пайплайн** на тех же данных:
   ```R
   devtools::load_all()
   gp(args)
   ```

2. **Сравнить метрики** по каждой модели:
   - Correlation (r)
   - RMSE
   - MAE
   - R²

3. **Проверить допуски** (tolerance):
   - |corr_py - corr_R| < 0.02
   - |rmse_py - rmse_R| < 0.05
